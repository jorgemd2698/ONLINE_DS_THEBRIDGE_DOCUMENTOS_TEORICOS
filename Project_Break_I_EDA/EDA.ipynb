```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import sqlite3
```

```python
df_federados = pd.read_csv("./Data_sets/accidentes.csv")
df_a = df_federados.copy()
df_licencias = pd.read_csv("./Data_sets/licencias_2024.csv", low_memory=False)
df_l = df_licencias.copy()
df_a.info(max_cols = 100)
```

```python
df_l.info()
```

```python
# Unimos ambos dataframes conservando unicamente las personas que han tenido accidentes
# Como habrá columnas duplicadas, python las asigna con "_y", y lo eliminamos
df = pd.merge(df_a, df_l, on=["Nombre", "Apellido1"], how='inner')
df = df.loc[:, ~df.columns.str.endswith("_y")]
df.info()
```
