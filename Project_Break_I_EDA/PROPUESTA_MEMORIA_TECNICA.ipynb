# Memoria Técnica  
## Análisis Exploratorio y Siniestralidad de Federados

---

## 1. Introducción

El presente proyecto tiene como objetivo analizar la relación entre las características de los federados y la siniestralidad asociada a los seguros contratados. El estudio se centra en comprender cómo variables demográficas y contractuales influyen tanto en la frecuencia como en la gravedad de los siniestros, proporcionando información relevante para la toma de decisiones estratégicas por parte de la entidad aseguradora.

Para ello, se ha llevado a cabo un proceso completo de preparación de datos, análisis exploratorio y contraste de hipótesis, utilizando información procedente de licencias federativas y registros de siniestros. La correcta limpieza y estructuración de los datos ha sido un paso clave para garantizar la fiabilidad de los resultados obtenidos.

---

## 2. Contexto del proyecto y fuentes de datos

El análisis se basa en dos conjuntos de datos principales:

- **Dataset de licencias federativas**, que recoge información personal, demográfica y contractual de los federados.
- **Dataset de siniestros**, que contiene información detallada sobre los accidentes registrados, su gravedad y características asociadas.

Ambos datasets presentan una estructura heterogénea y requieren un proceso de limpieza y normalización previo a su análisis conjunto.

---

## 3. Hipótesis planteadas

Al inicio del estudio se formularon las siguientes hipótesis:

### Hipótesis A  
Existen diferencias significativas en la gravedad de los siniestros según la modalidad de seguro contratada.

### Hipótesis B  
Los federados mayores de 35 años presentan una gravedad media de siniestros superior a la de aquellos asociados a seguros básicos.

---

## 4. Preprocesado y limpieza de datos

### 4.1 Limpieza del dataset de licencias (Notebook 1)

El objetivo principal de esta fase fue construir un dataset analítico consistente y alineado con los objetivos del estudio. Las acciones realizadas incluyen:

- Tratamiento de valores faltantes en variables relevantes.
- Conversión de tipos de datos, especialmente fechas y variables categóricas.
- Normalización de variables categóricas como el sexo.
- Creación de variables derivadas, destacando la variable **edad**.
- Eliminación de columnas administrativas, bancarias y de contacto no relevantes para el análisis.

Como resultado, se obtuvo un dataset limpio, coherente y enfocado en el análisis del perfil del federado.

---

### 4.2 Limpieza del dataset de siniestros (Notebook 2)

El dataset de siniestros fue tratado para garantizar la calidad de la información relacionada con los accidentes. Las principales acciones fueron:

- Identificación y tratamiento de valores faltantes.
- Normalización de variables categóricas, especialmente la gravedad del siniestro.
- Conversión correcta de variables temporales.
- Eliminación de columnas no relevantes para el análisis de siniestralidad.

Este proceso permitió obtener un conjunto de datos robusto y preparado para el análisis estadístico.

---

## 5. Análisis univariante

### 5.1 Distribución de la gravedad de los siniestros

El análisis univariante muestra que la mayoría de los siniestros registrados son de carácter **leve**, mientras que los siniestros graves y mortales representan una proporción significativamente menor.

Este patrón es consistente con el tipo de actividad asegurada y proporciona un marco de referencia adecuado para los análisis posteriores.

---

### 5.2 Distribución de modalidades de seguro

La modalidad **B**, correspondiente a cobertura nacional, concentra la mayor parte de los federados. El resto de modalidades presenta una representación considerablemente menor, lo que condiciona la interpretación de los resultados comparativos.

---

### 5.3 Distribución de licencias por sexo

El análisis descriptivo muestra diferencias en el número de licencias contratadas según el sexo del federado. Este resultado aporta contexto demográfico al estudio, aunque no implica necesariamente diferencias en la siniestralidad.

---

## 6. Análisis bivariante

### 6.1 Modalidad de seguro y gravedad del siniestro

El cruce entre modalidad y gravedad del siniestro evidencia que, aunque existen diferencias en valores absolutos, la distribución relativa de la gravedad es similar entre modalidades, predominando los siniestros leves en todos los casos.

---

### 6.2 Edad y gravedad del siniestro

El análisis bivariante sugiere una ligera mayor presencia de siniestros graves en federados de mayor edad. No obstante, esta observación requiere validación mediante contraste estadístico.

---

## 7. Contraste de hipótesis

### 7.1 Hipótesis A: Modalidad y gravedad

Se aplicó una prueba estadística no paramétrica adecuada para contrastar diferencias entre grupos. El resultado indica que **no existen diferencias estadísticamente significativas** en la gravedad de los siniestros según la modalidad de seguro.

**Conclusión:**  
La hipótesis A queda **refutada**.

---

### 7.2 Hipótesis B: Edad y gravedad

El contraste estadístico realizado muestra que **no se detectan diferencias significativas** en la gravedad media de los siniestros entre federados mayores de 35 años y aquellos asociados a seguros básicos.

**Conclusión:**  
La hipótesis B queda **refutada**.

---

## 8. Síntesis de hallazgos

- Predominio claro de siniestros leves.
- Alta concentración de federados en la modalidad B.
- Ausencia de diferencias estadísticamente significativas en gravedad por modalidad o edad.
- El análisis por sexo aporta información descriptiva, sin impacto directo en la siniestralidad.

---

## 9. Conclusiones generales

El análisis exploratorio y estadístico realizado permite concluir que la gravedad de los siniestros no depende de forma significativa ni de la modalidad del seguro ni de la edad del federado. Esto sugiere una homogeneidad en el comportamiento del riesgo dentro del colectivo analizado.

---

- Distribución desigual entre modalidades.
- Presencia de valores no informados en algunas variables.
- Alcance temporal limitado a los datos disponibles.
